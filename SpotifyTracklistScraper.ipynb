```python
%pip install spotifyscraper
#Go to playlist and ctrl+a -> ctrl+c -> ctrl+v into notepad and save as "spotify_urls.txt"



import os
from spotify_scraper import SpotifyClient

INPUT_FILE = "spotify_urls.txt"
OUTPUT_FILE = "tracklist.txt"

# Make sure your file is there
if not os.path.exists(INPUT_FILE):
    print(f"❌ Error: Could not find '{INPUT_FILE}'.")
    exit()

print(f"Reading URLs and fetching song details anonymously...\n")

# Use SpotifyClient to automatically handle the anonymous connection
with SpotifyClient() as client, open(INPUT_FILE, "r", encoding="utf-8") as infile, open(OUTPUT_FILE, "w", encoding="utf-8") as outfile:
    
    for line in infile:
        url = line.strip()
        if not url:
            continue
            
        try:
            # The library accepts the full URL directly
            track = client.get_track(url)
            
            # Extract track name and primary artist
            song_title = track.name
            artists = ", ".join([artist.name for artist in track.artists])
            
            formatted_track = f"{song_title} - {artists}"
            print(f"🎵 {formatted_track}")
            
            outfile.write(formatted_track + "\n")
            
        except Exception as e:
            print(f"❌ Error fetching {url}: {e}")
            outfile.write(f"ERROR: Could not fetch {url}\n")

print(f"\n✅ Done! Saved to '{OUTPUT_FILE}'.")
```
